# Análise de Resíduos

```python
sns.histplot(df_validos["residuo"], kde=True)
plt.title("Distribuição dos Resíduos")
plt.show()

# df_validos["residuo"] -> seleciona a coluna chamada residuo do DataFrame | cada valor é um número que representa o erro da regressão
# sns.histplot(...) -> cria um histograma | cada barra representa a quantidade de resíduos em um intervalo
# kde=True -> adiciona uma curva suave (Kernel Density Estimation) | ajuda a visualizar a forma geral da distribuição

# Explicação : Isso tudo serve para entender se os resíduos estão centrados em zero,ver se há muitos resíduos negativos extremos,identificar assimetria, que sugere defeitos recorrentes
```

# APRENDIZADO SUPERVISIONADO

```python
limiar = df_validos["resistencia"].quantile(0.20)

# .quantile(0.20) -> retorna o valor abaixo do qual estão 20% dos dados, é o percentil 20
# limiar -> variável que guarda esse valor numérico


# Explicações : Isso mostra que 20% das peças têm resistência menor que esse valor,cria um critério objetivo para “resistência baixa”

df_validos["defeito"] = (df_validos["resistencia"] < limiar).astype(int)

# df_validos["resistencia"] < limiar -> gera uma série booleana : True se a resistência é menor que o limiar,False caso contrário
# .astype(int) -> converte para True -> 1 False -> 0
# df_validos["defeito"] = -> salva o resultado como nova coluna

#Explicações : Isso tudo para 1 = peça considerada defeituosa e 0 = peça considerada normal
```

```python
df_falha_total["defeito"] = 1

# cria a coluna "defeito" no DataFrame "df_falha_total" e atribui valor 1 para todas as linhas
# Para que assim toda falha total é automaticamente um defeito confirmado
```

# Preparação dos dados para ML

```python
features = [
    "diametro",
    "altura",
    "massa",
    "densidade_aparente"
]

# Aqui criamos uma lista de strings, onde cada string é o nome de uma coluna, funciona como um "manual" de variáveis para a entrada no modelo

X = df_ml[features] # seleciona apenas as colunas listadas e vira a matriz de entrada do modelo
y = df_ml["defeito"] # vira o vetor de saída (rótulo)
```

```python
X = X.fillna(X.median())

# X.median() -> calcula a mediana de cada coluna e retorna uma série com uma mediana por variável
# X.fillna(...) -> substitui valores NaN, assim cada coluna recebe sua própria mediana

# Utilizamos mediana por ser menos sensível a outliers e mais robusta que a média
```

# Treinamento supervisionado

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# test_size=0.3 -> É um parâmetro onde 30% dos dados vão para teste
# random_state=42 -> É um parâmetro que garante que a divisão seja sempre a mesma 
# Explicação: Divide os dados em dois conjuntos, 70% treino, 30% teste
```

```python
# SVM
svm = SVC()
svm.fit(X_train, y_train)

# SVC() -> Cria um classificador SVM com parâmetros padrão
# .fit(X_train, y_train) -> Ajusta o modelo aos dados e prende padrões que separam defeito e não defeito

y_pred = svm.predict(X_test)
print(classification_report(y_test, y_pred))

# .predict(X_test) -> Gera previsões para dados nunca vistos
# classification_report -> mostra: precisão, recall, f1-score, suporte
```

```python
# Árvore de Decisão

dt = DecisionTreeClassifier(max_depth=4, random_state=42)
dt.fit(X_train, y_train)

# max_depth=4 -> Limita o tamanho da árvore, evitando overfitting (quando um modelo de inteligência artificial ou estatística aprende "bem demais" os dados de treinamento, decorando detalhes e ruídos (erros aleatórios), em vez de aprender o padrão geral. )
# .fit(...) -> aprende regras do tipo: se densidade < X e massa < Y → defeito

y_pred_dt = dt.predict(X_test)
print(classification_report(y_test, y_pred_dt))
```

# Análise de Estabilidade dos Outliers (Validação Bootstrap)

## Contexto

Como o modelo lida com falsos positivos, ou seja,
se alguma amostra sem defeito intencional poderia ter sido classificada como outlier apenas
por variações normais do processo de impressão 3D.

Como não existe um gabarito real de quais amostras possuem defeito (a informação da fabricação
não foi disponibilizada à equipe de análise), não é possível calcular diretamente a taxa de
falsos positivos. Para tratar essa limitação de forma honesta e tecnicamente defensável, foi
realizada uma **análise de estabilidade via bootstrap**.

**Lógica da análise:** ao invés de treinar o Isolation Forest uma única vez, o modelo é
retreinado 200 vezes, cada vez sobre uma reamostragem bootstrap (com reposição) dos dados.
Uma amostra que é *realmente* anômala tende a ser marcada como outlier na maioria das rodadas,
independente de qual subconjunto de dados foi sorteado para treino. Já uma amostra que só foi
marcada por acaso (sensibilidade do modelo a variações normais do processo) tende a alternar
de forma inconsistente entre outlier/normal ao longo das rodadas.

Isso não substitui um gabarito real, mas fornece uma medida objetiva de confiança/robustez
para cada outlier detectado.

```python
# Parâmetros da análise de estabilidade

N_ITERACOES = 200          # número de reamostragens bootstrap
RANDOM_STATE_BASE = 42     # mesma semente usada na análise original, como base

# Mantemos o mesmo contamination usado na análise original (0.15),
# para que os resultados sejam comparáveis com o modelo já publicado
CONTAMINATION = 0.15

n_amostras = len(df_ml)

# Matriz para armazenar os resultados: uma linha por amostra, uma coluna por iteração
# Valor 1 = foi classificada como outlier naquela rodada; 0 = não foi
resultados_bootstrap = np.zeros((n_amostras, N_ITERACOES))

print(f"Iniciando análise de estabilidade ({N_ITERACOES} iterações bootstrap)...")
```

```python
# Loop de reamostragem bootstrap

for i in range(N_ITERACOES):

    # Sorteia índices com reposição (bootstrap) do mesmo tamanho do dataset original
    indices_bootstrap = np.random.RandomState(RANDOM_STATE_BASE + i).choice(
        n_amostras, size=n_amostras, replace=True
    )

    X_bootstrap = df_ml[features].iloc[indices_bootstrap]

    # Normalização (ajustada apenas nos dados da reamostragem)
    scaler_boot = StandardScaler()
    X_bootstrap_scaled = scaler_boot.fit_transform(X_bootstrap)

    # Treina o Isolation Forest na reamostragem, com o mesmo contamination original
    modelo_boot = IsolationForest(
        contamination=CONTAMINATION,
        random_state=RANDOM_STATE_BASE + i
    )
    modelo_boot.fit(X_bootstrap_scaled)

    # Aplica o modelo treinado sobre o dataset ORIGINAL completo (não só a reamostragem)
    # Isso permite avaliar todas as amostras originais em cada rodada, mesmo as que
    # não foram sorteadas para o treino daquela iteração
    X_original_scaled = scaler_boot.transform(df_ml[features])
    predicoes_originais = modelo_boot.predict(X_original_scaled)

    # Registra 1 onde foi classificado como outlier (-1) nesta rodada
    resultados_bootstrap[:, i] = (predicoes_originais == -1).astype(int)

print("Análise concluída.")
```

```python
# Cálculo da frequência de detecção como outlier (estabilidade)

freq_outlier = resultados_bootstrap.mean(axis=1)  # proporção de rodadas em que foi outlier

df_ml["frequencia_outlier"] = freq_outlier
df_ml["estabilidade_pct"] = (freq_outlier * 100).round(2)

# Classificação da robustez de cada amostra:
#   >= 90%  -> outlier robusto (alta confiança)
#   50-90%  -> outlier moderado (fronteira, requer cautela)
#   < 50%   -> não é consistentemente outlier (provável falso positivo pontual)

def classificar_robustez(freq):
    if freq >= 0.90:
        return "Robusto (alta confiança)"
    elif freq >= 0.50:
        return "Moderado (fronteira)"
    else:
        return "Baixa consistência"

df_ml["robustez_outlier"] = df_ml["frequencia_outlier"].apply(classificar_robustez)
```

```python
# Resultados: apenas amostras que foram outlier na análise original (anomalia == -1)

outliers_originais = df_ml[df_ml["anomalia"] == -1].copy()
outliers_originais = outliers_originais.sort_values("frequencia_outlier", ascending=False)

print("Estabilidade dos outliers detectados na análise original (2024):")
outliers_originais[["ID", "massa", "densidade_aparente", "estabilidade_pct", "robustez_outlier"]]
```

```python
# Resumo quantitativo

n_robustos = (outliers_originais["robustez_outlier"] == "Robusto (alta confiança)").sum()
n_moderados = (outliers_originais["robustez_outlier"] == "Moderado (fronteira)").sum()
n_baixa_consistencia = (outliers_originais["robustez_outlier"] == "Baixa consistência").sum()
total_outliers = len(outliers_originais)

print(f"Total de outliers identificados na análise original: {total_outliers}")
print(f"  - Robustos (>= 90% das rodadas):     {n_robustos} ({n_robustos/total_outliers*100:.1f}%)")
print(f"  - Moderados (50-90% das rodadas):    {n_moderados} ({n_moderados/total_outliers*100:.1f}%)")
print(f"  - Baixa consistência (< 50%):        {n_baixa_consistencia} ({n_baixa_consistencia/total_outliers*100:.1f}%)")
```

```python
# Visualização: estabilidade de cada amostra outlier

plt.figure(figsize=(12, 6))

cores = outliers_originais["robustez_outlier"].map({
    "Robusto (alta confiança)": "darkred",
    "Moderado (fronteira)": "orange",
    "Baixa consistência": "gold"
})

plt.bar(
    outliers_originais["ID"].astype(str),
    outliers_originais["estabilidade_pct"],
    color=cores,
    edgecolor="black"
)

plt.axhline(90, color="darkred", linestyle="--", linewidth=1, alpha=0.7, label="Limiar Robusto (90%)")
plt.axhline(50, color="orange", linestyle="--", linewidth=1, alpha=0.7, label="Limiar Moderado (50%)")

plt.title("Estabilidade dos Outliers Detectados (Análise Bootstrap) - Base 2024")
plt.xlabel("ID da Amostra")
plt.ylabel("% de Rodadas Classificada como Outlier")
plt.xticks(rotation=90, fontsize=7)
plt.legend()
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()

# Interpretação: barras próximas de 100% = outliers robustos (baixo risco de falso positivo)
# Barras próximas do limiar de 50% ou abaixo = maior risco de terem sido marcadas por acaso
```

```python
# Visualização complementar: todas as amostras (não só os outliers originais)

plt.figure(figsize=(12, 6))

sns.scatterplot(
    x=df_ml["massa"],
    y=df_ml["densidade_aparente"],
    hue=df_ml["frequencia_outlier"],
    palette="Reds",
    s=80,
    edgecolor="black"
)

plt.title("Frequência de Detecção como Outlier (Todas as Amostras) - Base 2024")
plt.xlabel("Massa (g)")
plt.ylabel("Densidade Aparente")
plt.legend(title="Freq. Outlier", bbox_to_anchor=(1.05, 1), loc="upper left")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
